```python
# notebooks - for experiments, analysis, charts, explanations
import pandas as pd
import numpy as np
#prices contains Close,high,low,open,volume of the 30 stocks

prices = pd.read_csv(
    "../data/processed/stock_prices_clean.csv",
    parse_dates=["Date"]
)

prices.head()
```

```python
prices.info() #2897 trading days (11.5 yrs of data), 30 stocks
```

```python
#Create stock closing prices only 
close_columns = [] #to store close columns 
for col in prices.columns:  #prices contains Close,high, low, open,volume
    if "Close" in col:
        close_columns.append(col) #['Close_AAPL', 'Close_AMZN', 'Close_AVGO', ....]
close_prices = prices[close_columns]
print(close_prices.head())
```

```python
close_prices.shape #(2897, 30), 2897 rows = trading days, 30 columns = stocks 
```

```python
#Calculate Daily returns 
#Convert prices into percentage changes - return = (today price - yesterday price) / yesterday price
apple_price_head = prices["Close_AAPL"].head()
print(apple_price_head)
#Convert prices into percentage changes - day 1 = 100, day 2 = 105, percentage change = 5%
apple_pct_head = apple_price_head.pct_change()
print("pct head change: \n")
print(apple_pct_head)
```

```python
#Calculate Daily returns these are the percentage changes
#Convert prices into percentage changes - return = (today price - yesterday price) / yesterday price
#Convert prices into percentage changes - day 1 = 100, day 2 = 105, percentage change = 5%
close_prices_percentage_change = close_prices.pct_change()
print(close_prices_percentage_change.head())
```

```python
#Remove any rows with NaN values (first row will be NaN because there is no previous day to compare to)
close_prices_percentage_change = close_prices_percentage_change.dropna()
print(close_prices_percentage_change.head())
print("Shape of close_prices_percentage_change: ", close_prices_percentage_change.shape) #(2896, 30)
```

```python
#Momentum factor
# Measures how much each stock has increased or decreased over the past 6 months (126 trading days).
#Formula
#Momentum = (Price today / Price 126 trading days ago) - 1
#Higher momentum = stronger recent performance, lower momentum = weaker recent performance
momentum = close_prices.pct_change(periods=126) #126 trading days in 6 months
print(momentum)
momentum.to_csv("../data/processed/momentum.csv")
```

```python
#calculate the volatility 
#measures how much do this stock's daily returns bounce around - standard deviation
volatility = close_prices_percentage_change.rolling(20).std() 
print(volatility)
print("Shape of volatility: ", volatility.shape) #(30,) 30 stocks, 1 value for each stock
```

```python
#Calculate the 200 day moving average 
# A moving average smooths out daily price noise, find the average price over the last x trading days 
# 200 trading days approx 1 year 
# Current price > 200 DMA , long term uptrend
# Current price < 200 DMA , long term downtrend
moving_average_200 = close_prices.rolling(window=200).mean() #rolling window looks at the previous 200 trading days, mean calculates the average
print("Moving average: ", moving_average_200)
print("Moving average head: ", moving_average_200.head())
moving_average_200.to_csv("../data/processed/moving_average_200.csv")   
```

```python
#Trend factor
#is the stock currently trading above or below its long term trend 
# compare current price vs 200 day moving average
# Rule: if current price > 200 day moving average -> uptrend = 1
# if current price < 200 day moving average -> downtrend = 0
trend = close_prices > moving_average_200
#Convert True/false to 1/0
trend_factor = trend.astype(int)
# print(trend)
print(type(momentum))
print(type(volatility))
print(type(trend_factor))

print(momentum.shape)
print(volatility.shape)
print(trend_factor.shape)
```

```python
#combine all factors into a single dataframe (momentum, volatility, trend_factor)
# print("Shape of momentum: ", momentum.shape) #(2897, 30)
# print("Shape of volatility: ", volatility.shape) #(30,)
# print("Shape of trend_factor: ", trend_factor.shape) #(2897, 30)

# Rename each DataFrame's columns to include the factor name
new_columns = []
for col in momentum.columns:
    if col.endswith("_Momentum"):
        new_columns.append(col)
    else:
        new_columns.append(col + "_Momentum")
momentum.columns = new_columns

new_columns = []
for col in volatility.columns:
    if col.endswith("_Volatility"):
        new_columns.append(col)
    else:
        new_columns.append(col + "_Volatility")
volatility.columns = new_columns

new_columns = []
for col in trend_factor.columns:
    if col.endswith("_Trend"):
        new_columns.append(col)
    else:
        new_columns.append(col + "_Trend")
trend_factor.columns = new_columns

#concatenate factors horizontally 
factor_scores = pd.concat([momentum,volatility,trend_factor],axis=1)

#check result
print("Momentum shape:", momentum.shape)
print("Volatility shape:", volatility.shape)
print("Trend shape:", trend_factor.shape)

print("Factor scores shape:", factor_scores.shape)

print(factor_scores.isna().sum().sum())
print(factor_scores.head())

#save
factor_scores.to_csv("../data/processed/new_factor_scores.csv")

#Remove any rows with NaN values
factor_scores = factor_scores.dropna()
print(factor_scores.shape)
print(factor_scores.isna().sum().sum())
factor_scores.to_csv("../data/processed/factor_scores_clean.csv")
```

```python
#Normalize the factors into a standardized score (z-score)
#Mean = 0, standard deviation = 1
#Composite score = 50% Momentum, 30% Volatility, 20% Trend
#Z score normalization, z = (x-mean)/standard deviation
#positive number -> above average, negative number -> below average, 0 -> exactly average

#Separate factor groups
#Get all momentum columns
momentum_columns=[]

for col in factor_scores.columns:
    if "Momentum" in col:
        momentum_columns.append(col)
print("Momentum columns: ", momentum_columns)
print("Number of momentum columns: ", len(momentum_columns))

#Get all volatility columns
volatility_columns=[]
for col in factor_scores.columns:
    if "Volatility" in col:
        volatility_columns.append(col)
print("Volatility columns: ", volatility_columns)
print("Number of volatility columns: ", len(volatility_columns))

#Get all trend columns
trend_columns=[]
for col in factor_scores.columns:
    if "Trend" in col:
        trend_columns.append(col)
print("Trend columns: ", trend_columns)
print("Number of trend columns: ", len(trend_columns))


```

```python
#Extract only momentum data
momentum_data = factor_scores[momentum_columns]
print("Finished extracting momentum data ")
print("Momentum data shape: ", momentum_data.shape)

#calculate momentum mean
momentum_mean = momentum_data.mean()
print("Momentum mean: ", momentum_mean.head())

#calculate momentum standard deviation
momentum_std = momentum_data.std()
print("Momentum std: ", momentum_std.head())

#calculate z score
#z = (x - mean) / standard deviation
momentum_normalized = (momentum_data - momentum_mean) / momentum_std
# print("Momentum normalized:")
print(momentum_normalized.head())
print(momentum_data.shape)
```

```python
#Normalize the volatility factor 
#extract only the volatility data 
volatility_data = factor_scores[volatility_columns]
print("Volatility data shape: ", volatility_data.shape)

#calculate volatility mean
volatility_mean = volatility_data.mean()
print("Volatility mean: ", volatility_mean.head())

#calculate volatility standard deviation
volatility_std = volatility_data.std()
print("Volatility std: ", volatility_std.head())

#calculate volatility z score
volatility_normalized = (volatility_data - volatility_mean) / volatility_std
print("Volatility normalized: ")
print(volatility_normalized.head())
```

```python
#Normalize the trend factor
#extract only the trend data
trend_data = factor_scores[trend_columns]
print("Trend data shape: ", trend_data.shape)
#calculate trend mean
trend_mean = trend_data.mean()
print("Trend mean: ", trend_mean.head())

#calculate trend standard deviation
trend_std = trend_data.std()
print("Trend std: ", trend_std.head())

#calculate trend z score
trend_normalized = (trend_data - trend_mean) / trend_std
print("Trend normalized: ")
print(trend_normalized.head())
```

```python
print("Momentum normalized shape: ", momentum_normalized.shape)
print("Volatility normalized shape: ", volatility_normalized.shape)
print("Trend normalized shape: ", trend_normalized.shape)
```

```python
#find which factor has NaN values
print("Momentum NaNs:")
print(momentum_normalized.isna().sum().sum())

print("Trend NaNs:")
print(trend_normalized.isna().sum().sum())

print("Volatility NaNs:")
print(volatility_normalized.isna().sum().sum())

print(momentum_normalized.columns[:5])
print(trend_normalized.columns[:5])
print(volatility_normalized.columns[:5])

# Remove factor suffixes so columns match across factors

momentum_normalized.columns = (
    momentum_normalized.columns
    .str.replace("_Momentum", "", regex=False)
)

trend_normalized.columns = (
    trend_normalized.columns
    .str.replace("_Trend", "", regex=False)
)

volatility_normalized.columns = (
    volatility_normalized.columns
    .str.replace("_Volatility", "", regex=False)
)

print(momentum_normalized.columns[:5])
print(trend_normalized.columns[:5])
print(volatility_normalized.columns[:5])
```

```python
#Decide factor weights 
#Momentum = 50%,  Trend = 30% Volatility = 20%,
#create composite score by multiplying each factor by its weight and summing them up
composite_score = (
    0.5 * momentum_normalized +
    0.3 * trend_normalized -
    0.2 * volatility_normalized
)
print("comosite score. created")
print(composite_score)

print("Composite score shape: ", composite_score.shape)
```
